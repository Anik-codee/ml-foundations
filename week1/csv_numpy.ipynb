```python
import csv                               #Reading CSV file and printing machines with temperature higher than 75
with open('sensor_data.csv','r') as f:
    reader= csv.DictReader(f)
    for row in reader:
        if int(row['temperature']) > 75:
            print(f"Machines with temperature greater than 75 are {row['machine']} with temperature: {row['temperature']}")
```

```python
import numpy as np                  #Basic NumPy operations
temperatures = np.array([68, 82, 74, 95, 65])
print(temperatures.shape)
print(f"Mean of the temperatures is: {np.mean(temperatures)}")
print(f"Standard deviation is: {np.std(temperatures)}")
print(f"Minimum value is: {np.min(temperatures)}")
print(f"Maximum value is: {np.max(temperatures)}")
print(f"Median is: {np.median(temperatures)}")
greater_values=temperatures[temperatures>75]
print(greater_values)
positions=np.where(temperatures>75)
print(positions)
```
